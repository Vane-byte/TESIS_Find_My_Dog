```python
import tensorflow as tf
import numpy as np
```

```python
with open ('datos_tokenizados/23-10-2022/posts-xids.npy', 'rb') as f:
     Xids = np.load(f,allow_pickle=True)
with open ('datos_tokenizados/23-10-2022/posts-xmask.npy', 'rb') as f:
     Xmask = np.load(f,allow_pickle=True)
with open ('datos_tokenizados/23-10-2022/posts-labels.npy', 'rb') as f:
     labels = np.load(f,allow_pickle=True)
labels=labels.tolist()
```

```python
dataset= tf.data.Dataset.from_tensor_slices((Xids, Xmask, labels))
dataset.take(1)
```

```python
#inputs, outputs
def map_func(input_ids,masks,labels):
     return({'input_ids':input_ids,
               'attention_mask':masks}, labels)
dataset= dataset.map(map_func)
dataset.take(1)
```

DATASET SHUFFLE, BATCH SPLIT AND SAVE

```python
batch_size= 16
dataset=dataset.shuffle(100).batch(batch_size,drop_remainder=True)
dataset.take(1)
```

```python
split= 0.9
size = int(Xids.shape[0]/batch_size * split)
size
```

```python
train_ds= dataset.take(size)
val_ds= dataset.skip(size)
tf.data.experimental.save(train_ds,'dataset/22-02-26/train')
tf.data.experimental.save(val_ds,'dataset/22-02-26/val')
```
