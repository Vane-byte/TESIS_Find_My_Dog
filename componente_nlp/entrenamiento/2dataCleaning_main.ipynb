```python
import pandas as pd
pd.set_option('display.max_colwidth', None)
```

```python
df= pd.read_csv('datos_sin procesar/extracc_twitter_4.csv',';')
df= pd.DataFrame(df['text'])

#df= pd.read_csv('datos_sin procesar/data_preproc_todo.csv', sep=';')
#df= df.drop(columns='Unnamed: 0')
```

```python
import re
from cleantext import clean

for row in df.itertuples():
    df.at[row.Index, "text"] = re.sub(r'http\S+', '', (df.at[row.Index, "text"])) # no links
    df.at[row.Index, "text"] = re.sub(r'\n','. ', (df.at[row.Index, "text"])) # no \n
    df.at[row.Index, "text"] = clean(df.at[row.Index, "text"],to_ascii=False, no_emoji=True, normalize_whitespace=True, 
                                lower=False, lang='es') # no emojis o caracteres especiales
    df.at[row.Index, "text"] = re.sub(r'#', '', (df.at[row.Index, "text"]))
    df.at[row.Index, "text"] = re.sub(r'@\S+', 'USER_ID', (df.at[row.Index, "text"]))
```

```python
df=df.drop_duplicates()
df
```
